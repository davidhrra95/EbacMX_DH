```python
#Función lambda para obtener la raíz cuadrada de un número.
num = 10
raiz_cuadrada = num ** 0.5
print(raiz_cuadrada)

#Usando Lambda
raiz = lambda x: x ** 0.5
print(raiz(10))
```

```python
#Función map, para obtener el largo de una cadena de palabras. 

cadena = ['David', 'Hola', 'Cuatrocientos']
largo = []
for palabras in cadena:
    x = len(palabras)
    largo.append(x)
print(largo)

#Usando map
lista_map = list(map(len, cadena))
print(lista_map)
```

```python
#Función reduce, que sirva para calcular el producto de una lista 
from functools import reduce
lista = [1,2,3,4,5]
producto = 1
for x in lista:
    producto *= x
print(producto)

#Usando reduce
resultado = reduce((lambda a, b: a * b), lista)
print(resultado)
```

```python
#Función filter que sirva para encontrar palabras que contengan mayúsculas o números en un listado 
cadenaxfiltrar = ['David', 'hola', 'Cuatrocientos', '777', '32']
resultados = [palabra for palabra in cadenaxfiltrar
              if any(c.isupper() or c.isdigit()
                     for c in palabra)]
print(resultados)

#Usando filter
resultados_filter = list(filter(lambda palabra: any(c.isupper() or c.isdigit() for c in palabra), cadenaxfiltrar))
print(resultados_filter)
```
