```python
#Leer el archivo CSV mencionado 
import csv
archivo = open(r'/Users/davidherrera/Downloads/Height_of_Male_and_Female_by_Country_2022.csv')
csvreader = csv.reader(archivo)
print(csvreader)

encabezado = []
encabezado = next(csvreader)
print(encabezado)
```

```python
#Imprimir las primeras 50 líneas
rows = []
for row in csvreader:
    rows.append(row)
print('Columnas:\n',rows[:50])
```

```python
#Calcular la media, mínima y máxima altura de hombres y mujeres del data set completo
import numpy as np

tabla = np.array(rows)
print('Tabla:\n',tabla[:5],'\n')

alturah = tabla[:,2]
alturam = tabla[:,3]
print('Altura Hombres:\n',alturah[:5],'\n')
print('Altura Mujeres:\n',alturam[:5],'\n')

alturah_cmfloat = [float(n) for n in alturah]
alturam_cmfloat = [float(n) for n in alturam]

from statistics import mean

mean_alturah_cm = mean(alturah_cmfloat)
max_alturah_cm = max(alturah_cmfloat)
min_alturah_cm = min(alturah_cmfloat)
mean_alturam_cm = mean(alturam_cmfloat)
max_alturam_cm = max(alturam_cmfloat)
min_alturam_cm = min(alturam_cmfloat)

print('Altura promedio en Hombres: {:.2f} \nAltura maxima en Hombres: {} \nAltura minima en Hombres: {}'.format(mean_alturah_cm, max_alturah_cm, min_alturah_cm))
print('Altura promedio en Mujeres: {:.2f} \nAltura maxima en Mujeres: {} \nAltura minima en Mujeres: {}'.format(mean_alturam_cm, max_alturam_cm, min_alturam_cm))


```
