```python
#JSON -leer el archivo y guardarlo en una lista: 
import json

with open('season1.json', 'r', encoding='utf-8') as jason_file:
    s1 = json.load(jason_file)

key = 'Game Of Thrones S01E01 Winter Is Coming.srt'

if key in s1:
    data = s1[key]
```

```python
#JSON - obtener las 10 primeras filas:
first_10_rows = list(data.items())[:10]

for row in first_10_rows:
    print(row)
```

```python
#JSON - obtener las 10 últimas filas:
last_10_rows = list(data.items())[-10:]

for row in last_10_rows:
    print(row)
```

```python
#JSON - seleccionar 2-4 columnas del dataset y generar un archivo nuevo:
print(s1.keys())

select_keys = ['Game Of Thrones S01E01 Winter Is Coming.srt', 'Game Of Thrones S01E02 The Kingsroad.srt']

got_nvo = {k: v for k, v in s1.items() if k in select_keys}

with open('got_nuevo.json', 'w', encoding='utf-8') as f:
    json.dump(got_nvo, f, ensure_ascii=False, indent=4)
```

```python
#EXCEL - leer el archivo y guardarlo en una lista: 
import openpyxl
from pathlib import Path

xlsx_file = Path('.', 'Amazon_top100_bestselling_books_2009to2021.xlsx')
wb_obj = openpyxl.load_workbook(xlsx_file)
sheet = wb_obj.active

print(sheet.max_row, sheet.max_column)
```

```python
#EXCEL - obtener las 10 primeras filas:
first_10 = sheet['A1':'J10']
print(first_10)
```

```python
#EXCEL - obtener las 10 últimas filas:
last_10 = sheet['A1282':'J1292']
print(last_10)
```

```python
#EXCEL - seleccionar 2-4 columnas del dataset y generar un archivo nuevo:
nuevo_wb = openpyxl.Workbook()
nueva_hoja = nuevo_wb.active
columnas_deseadas = [2, 4]

for fila in sheet.iter_rows(values_only=True):
    nueva_fila = [fila[i-1] for i in columnas_deseadas]
    nueva_hoja.append(nueva_fila)

nuevo_wb.save('Amazon_columnas_seleccionadas.xlsx')
```
