```python
#Uso de librería BeautifulSoup y requests.  Parseo de información (e.g. mercadolibre, despegar.com, etc.) 
import requests
from bs4 import BeautifulSoup

url = 'https://listado.mercadolibre.com.mx/iphone#D[A:iphone]'

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

try:
    data = requests.get(url, headers=headers)
    data.raise_for_status()
    html = BeautifulSoup(data.text, 'lxml')
    items = html.select('.ui-search-layout__item')
    print(f"Se encontraron {len(items)} artículos en la página.\n")

    for item in items:
        print('-------------------------')

        try:
            title_element = item.select_one('.poly-component__title')
            if title_element:
                title = title_element.get_text(strip=True)
                print(title)
                price_element = item.select_one('.poly-price__current .andes-money-amount__fraction')
                if price_element:
                    print(f"Precio: ${price_element.get_text(strip=True)}")
            else:
                print('Título no encontrado.')
        except Exception as e:
            print(f'Error procesando el artículo: {e}')

except requests.exceptions.RequestException as e:
    print(f"Error al cargar la URL: {e}")
```
