```python
#Generar una lambda que añada la zona horaria y la capital del país de nacionalidad
import country_converter as coco

df_unido['CC'] = coco.convert(names=df_unido['Nationality'], to='ISO2', not_found=None)
```

```python
import pytz

df_unido['Zona Horaria'] = df_unido['CC'].apply(lambda x:pytz.country_timezones.get(x))
```

```python
from countryinfo import CountryInfo

def cap(pais):
    if pd.isna(pais):
        return pd.NA   
    try:
        return CountryInfo(pais).capital()
    except:
        return pd.NA

df_unido['Capital'] = df_unido['Nationality'].apply(cap)
```

```python
#Generar una lambda que en base al potencial, añada una columna “Candidato” o “Next Year”, para el premio para el balón de oro.  Jugadores con más de 85, son candidatos.#

df_unido['Candidato'] = df_unido['Potential'].apply(lambda x:'Yes' if x > 85 else 'No')
```
