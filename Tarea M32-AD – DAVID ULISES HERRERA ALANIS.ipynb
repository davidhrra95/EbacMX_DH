```python
#Generar un archivo tipo Notebook de Python que contenga el código fuente de 
#Leer archivo fifa_eda.csv  

import pandas as pd
import plotly.express as px
import os

os.chdir('/Users/davidherrera/Downloads')

df = pd.read_csv('fifa_eda.csv')

df.sample(3)
```

```python
#Elemento 1: Obtener matriz de correlación y dejarla como heat map 
corr = df[['Age', 'Potential', 'Value']]
fig = px.imshow(corr.corr(numeric_only=True), text_auto=True, title="Correlación entre Edad, Potencial y Valor")
fig.show()
```

```python
#Elemento 2: Gráfico de Correlación que responda – “Cuál es la relación entre la edad y el overall?” – explicar la salida 
corr = df[['Age', 'Overall']]
fig = px.imshow(corr.corr(numeric_only=True), text_auto=True, title="¿Cuál es la relación entre la edad y el overall?")
fig.show()
```

```python
#Elemento 3: Generar un gráfico de barras por club que indique el número de jugadores 
club = df.groupby(['Club']).size().rename('num_jugadores').reset_index()
fig = px.bar(club.head(10), x='Club', y='num_jugadores', title="Número de jugadores por club")
fig.show()
```

```python
#Elemento 4: Hacer un gráfico multipanel que indique la relación altura (height) vs skill moves, siendo la variable del panel si es zurdo o derecho 
fig = px.strip(df, x="Height", y="Skill Moves", facet_col="Preferred Foot")
fig.show()
```

```python
#Elemento 5: Un gráfico que crea importante para mostrar a los potenciales cracks
valor = df['Nationality'].value_counts().head(20)
fig = px.bar(valor, title="Top paises con más jugadores")
fig.show()
```
