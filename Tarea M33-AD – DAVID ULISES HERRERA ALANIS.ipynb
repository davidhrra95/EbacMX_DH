```python
#Elementos mínimos:
#Totales: 
#1) Total de videojuegos analizados.
total_registros = len(df)
print(f"Total de registros analizados: {total_registros}")
total_juegos_unicos = df['Name'].nunique()
print(f"Total de videojuegos únicos: {total_juegos_unicos}")
```

```python
#2) total de géneros analizados.
total_generos_unicos = df['Genre'].nunique()
print(f"Total de generos únicos: {total_generos_unicos}")
```

```python
#3) Plataformas analizadas
total_plataformas_unicos = df['Platform'].nunique()
print(f"Total de plataformas únicos: {total_plataformas_unicos}")
```

```python
#Ventas:
#1) Total de ventas por año.
df.groupby('Year')['Global_Sales'].sum().reset_index()
```

```python
#2) Juegos más vendidos por año.
top_juegos = df.sort_values(['Year', 'Global_Sales'], ascending=[True, False])
top_juegos_por_año = top_juegos.drop_duplicates(subset=['Year'])
top_juegos_por_año[['Year', 'Name', 'Global_Sales']].reset_index()
```

```python
#3) Plataformas con más ingresos por año. 
    #Todos estos números desglosados por ingreso:
    #NA, EU, JP, Otros y el Global.

top_plataformas = df.sort_values(['Year', 'Global_Sales'], ascending=[True, False])
top_plataformas_por_año = top_plataformas.drop_duplicates(subset=['Year'])
top_plataformas_por_año[['Year', 'Platform', 'NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']].reset_index()
```

```python
#4) Gráfico de comparación de ventas entre géneros por año.
df.groupby(['Year', 'Genre'])['Global_Sales'].sum().unstack().plot(title="Ventas entre generos por año");
```

```python
#Insights: 2 Gráficos adicionales interesantes:
df.groupby('Year')['Global_Sales'].sum().plot.bar(title="Total ventas por año");
```

```python
df.groupby('Genre')['Global_Sales'].sum().plot.bar(title='Ventas totales por genero');
```

```python
#Revisar el EDA del ejercicio anterior y añadir siguientes elementos, basados en los últimos 4 años
#Crecimiento:
#1) qué género ha tenido el mayor y menor crecimiento.
ventas_genero = df.groupby(['Year', 'Genre'])['Global_Sales'].sum().unstack().fillna(0)
ventas_genero = ventas_genero.replace(np.nan, 0)
ventas_genero.style.background_gradient(cmap='Greens').set_properties(**{'font-size':'10px'})
```

```python
#2) qué plataforma ha tenido el mayor y menor crecimiento,
ventas_plataforma = df.groupby(['Year', 'Genre'])['Global_Sales'].sum().unstack().fillna(0)
ventas_plataforma = ventas_plataforma.replace(np.nan, 0)
ventas_plataforma.style.background_gradient(cmap='Blues').set_properties(**{'font-size':'10px'})
```

```python
#3) qué videojuego ha tenido el mayor y menor crecimiento
# Agrupamos por nombre y año para consolidar todas las plataformas
dfsales = df.groupby(['Name','Year'])['Global_Sales'].sum().sort_values(ascending=False).reset_index()
dfsales.head(10)
```
